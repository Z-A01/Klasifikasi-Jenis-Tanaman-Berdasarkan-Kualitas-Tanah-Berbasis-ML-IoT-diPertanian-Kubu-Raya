Data Preprosesing

```python
#menghapus fitur
df = df.drop(columns=['No', 'ID Titik', 'Timestamp'])
```

```python
#Mising value
df.isnull().sum()
```

```python
#Pengecekan Duplicate
df.duplicated().sum()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

features = df.select_dtypes(include='number').columns.tolist()
features = [col for col in features if col not in ['No', 'ID Titik']]

hasil = []
for _, group in df.groupby('Label'):
    mask = pd.Series(True, index=group.index)
    for col in features:
        Q1, Q3 = group[col].quantile([0.25, 0.75])
        IQR = Q3 - Q1
        mask &= group[col].between(Q1 - 1.5*IQR, Q3 + 1.5*IQR)
    hasil.append(group[mask])

df_clean = pd.concat(hasil)

print("Sebelum :", len(df))
print("Sesudah :", len(df_clean))
print("Berkurang :", len(df) - len(df_clean))

order = sorted(df_clean['Label'].unique())
ax = sns.countplot(data=df_clean, x='Label', order=order, palette='Set2')
for bar in ax.patches:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height(), int(bar.get_height()), ha='center', va='bottom')
plt.title('Jumlah Data Setelah Penghapusan Outlier')
plt.xlabel('Kelas Tanaman')
plt.ylabel('Jumlah Data')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
hasil_ph = (
    df_clean.groupby('Label')['pH']
      .agg(
          Jumlah_Data='count',
          pH_Min='min',
          pH_Max='max'
      )
      .reset_index()
)

display(hasil_ph)
```

```python
# 6. LABEL ENCODING

from sklearn.preprocessing import LabelEncoder

df_encoded = df_clean.copy()

le = LabelEncoder()
df_encoded['Label'] = le.fit_transform(df_encoded['Label'])

mapping = pd.DataFrame({
    'Label Tanaman': le.classes_,
    'Kode': le.transform(le.classes_)
})

print("=== Tabel Label Encoding ===")
display(mapping)
```

```python
# Min-Max Scaler
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. Pisahkan Fitur (X) dan Label (y)
X = df_encoded.drop(columns=['Label'])
y = df_encoded['Label'].values

# 2. Proses Feature Scaling
scaler = MinMaxScaler()
X_scaled_array = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled_array, columns=X.columns)

# 3. Gabungkan Kembali Data Bersih
df_siap_modeling = X_scaled.copy()
df_siap_modeling['Label'] = y

# 4. Tampilkan Hasil (Menggunakan array hasil transformasi)
print(f"Min setelah scaling: {X_scaled_array.min():.2f}")
print(f"Max setelah scaling: {X_scaled_array.max():.2f}")
display(df_siap_modeling.head())
display(df_siap_modeling.tail())
```

```python
#Train-split
from sklearn.model_selection import train_test_split
import pandas as pd

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(f"Total: {len(X_scaled)} | Train: {len(X_train)} | Test: {len(X_test)}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

rf_model = RandomForestClassifier(max_depth=5, random_state=42)
nb_model = GaussianNB()

rf_model.fit(X_train, y_train)
nb_model.fit(X_train, y_train)

rf_train_acc = rf_model.score(X_train, y_train) * 100
nb_train_acc = nb_model.score(X_train, y_train) * 100

print("=== AKURASI TAHAP TRAINING ===")
print(f"Akurasi Training Random Forest : {rf_train_acc:.2f}%")
print(f"Akurasi Training Naïve Bayes   : {nb_train_acc:.2f}%\n")
```

```python
#report dan confusion matrix
rf_pred = rf_model.predict(X_test)
nb_pred = nb_model.predict(X_test)

print("=== RANDOM FOREST ===")
print(f"Akurasi: {accuracy_score(y_test, rf_pred) * 100:.2f}%")
print(classification_report(y_test, rf_pred))

print("\n=== NAÏVE BAYES ===")
print(f"Akurasi: {accuracy_score(y_test, nb_pred) * 100:.2f}%")
print(classification_report(y_test, nb_pred))

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

sns.heatmap(confusion_matrix(y_test, rf_pred), annot=True, fmt='d', cmap='Blues', ax=ax[0])
ax[0].set_title('Random Forest')

sns.heatmap(confusion_matrix(y_test, nb_pred), annot=True, fmt='d', cmap='Oranges', ax=ax[1])
ax[1].set_title('Naïve Bayes')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB

X_cv = X_scaled
y_cv = y

rf_model_cv = RandomForestClassifier(max_depth=5, random_state=42)
nb_model_cv = GaussianNB()

cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rf_scores = cross_val_score(rf_model_cv, X_cv, y_cv, cv=cv_strategy, scoring='accuracy')
nb_scores = cross_val_score(nb_model_cv, X_cv, y_cv, cv=cv_strategy, scoring='accuracy')

print("=== RANDOM FOREST ===")
for i, score in enumerate(rf_scores, 1):
    print(f"Fold {i}: {score * 100:.2f}%")
print(f"Rata-rata: {rf_scores.mean() * 100:.2f}%")
print(f"Std Dev: {rf_scores.std() * 100:.2f}%\n")

print("=== NAÏVE BAYES ===")
for i, score in enumerate(nb_scores, 1):
    print(f"Fold {i}: {score * 100:.2f}%")
print(f"Rata-rata: {nb_scores.mean() * 100:.2f}%")
print(f"Std Dev: {nb_scores.std() * 100:.2f}%")
```

```python
# ============================================================
# EVALUASI MODEL DENGAN STRATIFIED K-FOLD 5
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB


# ------------------------------------------------------------
# MODEL
# ------------------------------------------------------------

models = {
    'Random Forest': RandomForestClassifier(
        max_depth=5,
        random_state=42
    ),

    'Naive Bayes': GaussianNB()
}


# ------------------------------------------------------------
# K-FOLD 5
# ------------------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# METRIK
# ------------------------------------------------------------

scoring = {
    'Accuracy': 'accuracy',
    'Precision': 'precision_macro',
    'Recall': 'recall_macro',
    'F1-Score': 'f1_macro'
}


# ------------------------------------------------------------
# PROSES EVALUASI
# ------------------------------------------------------------

hasil = []

for nama, model in models.items():

    scores = cross_validate(
        model,
        X_scaled,
        y,
        cv=cv,
        scoring=scoring
    )

    hasil.append({
        'Model': nama,
        'Accuracy': scores['test_Accuracy'].mean(),
        'Precision': scores['test_Precision'].mean(),
        'Recall': scores['test_Recall'].mean(),
        'F1-Score': scores['test_F1-Score'].mean()
    })


# ------------------------------------------------------------
# TABEL HASIL
# ------------------------------------------------------------

hasil_df = pd.DataFrame(hasil)

# Ubah ke persen
kolom_metrik = [
    'Accuracy',
    'Precision',
    'Recall',
    'F1-Score'
]

hasil_df[kolom_metrik] = hasil_df[kolom_metrik] * 100

# Pembulatan 2 angka
hasil_df[kolom_metrik] = hasil_df[kolom_metrik].round(2)

# Tampilkan tabel
display(hasil_df)
```
